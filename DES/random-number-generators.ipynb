Given integer parameters $a, c, m$, a Linear Congruential Generator (LCG) is a recurrence equation:

$$
x_n = (ax_{n-1} + c) \mod m
$$

Starting from an initial _seed_ $x_0$, a LCG generates a sequence of integers in $\{0, \ldots, m-1\}$.

We can realize a generic LCG as a Python generator.

```python
def LCG(a, c, m, x0):
    x = x0
    while True:
        yield x
        x = (a*x + c) % m
```

Foe example, let us consider the recurrence $x_n = (5x_{n-1} +1) \mod 16$. It can be shown that this generator has cycle length 16 (the maximum possible), as we can empirically observe by picking a seed, say $x_0 = 0$.

```python
import itertools as it

r = LCG(5, 1, 16, 0)

print(list(it.islice(r, 32)))
```

LCGs are very simple and fast. However, they have a small state space and often are unable to achieve maximum cycle length. 

```python
import numpy as np

N = 10000
# generate N random numbers using Python's random()
rpython = np.random.default_rng(123).random(N)
# a, c, m, x0 = 3, 0, 31, 1
# generate N random numbers using LCG(65539, 0, 2**31)
a, c, m, x0 = 65539, 0, 2**31, 17
rlcg = list(it.islice(LCG(a, c, m, x0), N))
```

A simple way to check the "goodness" of a PRNG is through a spectral test. We start with a 2D spectral test, that works as follows. Let $x_0, x_1, \ldots$ be the sequence of random numbers that we want to test. We plot the points of coordinates $(x_n, x_{n+1})$, $n=0, \ldots$, i.e., $(x_0, x_1), (x_1, x_2), (x_2, x_3), \ldots$, and so on, and look whether the points appears to be "randomly scattered" over the plane (this can be made more formally precise).

```python
from matplotlib import pyplot as plt
plt.rcParams['figure.dpi'] = 200

plt.subplots(1, 2, figsize=(16,8))

plt.subplot(1,2,1)
plt.scatter(rpython[:-1], rpython[1:], s=0.5)
plt.title("Numpy RNG")
plt.xlabel("$x_n$")
plt.ylabel("$x_{n+1}$")

plt.subplot(1,2,2)
plt.scatter(rlcg[:-1], rlcg[1:], s=0.5)
plt.title(f"LCG (a={a}, c={c}, m={m})")
plt.xlabel("$x_n$")

plt.show()
```

It appears that, in two dimensions, $\text{LCG}(65539, 0, 2^{31})$ is "as random as" Python default RNG. The spectral test can be performed in any number of dimensions: the $d$-dimensional spectral test consists of checking whether the $d$-dimensional points $(x_n, x_{n+1}, \ldots, x_{n+d-1})$, $n=0, 1, \ldots$, are randomly distributed over $d$-dimensional space. Let us try with three dimensions.

```python
ax = plt.figure(figsize=(10,10)).add_subplot(projection="3d")
plt.plot(rlcg[:-2], rlcg[1:-1], rlcg[2:], '.', markersize=0.5)
#ax.view_init(azim=65)
plt.title(f"LCG (a={a}, c={c}, m={m})")
ax.set_xlabel("$x_n$")
ax.set_ylabel("$x_{n+1}$")
ax.set_zlabel("$x_{n+2}$")
plt.show()
```

The points appear to be randomly distributed indeed. However, try to change che viewpoint by uncommenting the instruction `ax.view_init(azim=65)`...
